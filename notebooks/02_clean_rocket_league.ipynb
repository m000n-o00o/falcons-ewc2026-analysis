# 02 - Cleaning Rocket League data

Checks and cleans the raw EWC 2026 Falcons data collected in notebook 01:
missing values, duplicates, column data types, and team name consistency.

```python
import pandas as pd

# Load the raw Rocket League data collected in notebook 01
csv_path = "../data/raw/rocket_league/ewc2026_falcons_matches.csv"
matches = pd.read_csv(csv_path)

print(f"Loaded {len(matches)} rows")
```

```python
## Check for missing values



pd.isnull(matches).sum()
```

```python
## Check column data types



matches.info()
```

```python
## Check for duplicate rows



matches.duplicated().sum()
```

```python
## Convert match_date from text to datetime



matches["match_date"] = pd.to_datetime(matches["match_date"])
```

```python
## Confirm match_date is now datetime instead of str



matches.info()
```

```python
## Check team name consistency



matches["team_name"].unique()
```

```python
## Save the cleaned data to data/processed



matches.to_csv("../data/processed/rocket_league/ewc2026_falcons_cleaned.csv", index=False)
```
